```python
import json
import uuid
from flask_bcrypt import Bcrypt
from datetime import datetime, timedelta
import random
import pandas as pd

bcrypt = Bcrypt()

first_names = ['John', 'Jane', 'Alice', 'Bob', 'Charlie']
last_names = ['Doe', 'Smith', 'Johnson', 'Williams', 'Brown']

# Sample roles
roles = ['TM']

def random_string(length=10):
    letters = 'abcdefghijklmnopqrstuvwxyz'
    return ''.join(random.choice(letters) for _ in range(length))

# Generate data again
advocates_data = []

# Manager's IDs
manager_ids = [1454, 1353, 1376, 1495, 1004]

roles_description = {
    'SUPER_ADMIN': "Super Admin",
    'MGMT': "Admin.",
    "BE": "Business Enablement",
    "GM": "Global Manager",
    'TM': "Team Manager",
    'DM': "Duty Manager",
    'OM': "Operations Manager",
    'SME': "Subject Matter Experts",
    'TL': "Technical Lead",
    'T2': "Tier 2 Support Advocate",
    'SE': "Tenured Advocate",
    'NH': "New Hires",
    'VIS': "Visitors"
}


for i in range(50):  # Generate data for 10 advocates
    reports_to = random.choice(manager_ids)
    schedules = json.dumps({"2024": {"JANUARY": None, "FEBRUARY": None, "MARCH": None, "APRIL": None, "MAY": None, "JUNE": None, "JULY": "TWTFS", "AUGUST": None, "SEPTEMBER": None, "OCTOBER": None, "NOVEMBER": None, "DECEMBER": None}})
    id_ = str(uuid.uuid4())
    staff_id = random.randint(1000, 9999)
    created_at = datetime.now() - timedelta(days=random.randint(1, 100))
    updated_at = created_at + timedelta(days=random.randint(1, 100))
    first_name = random.choice(first_names)
    last_name = random.choice(last_names)
    status = True
    email = f'{first_name.lower()}.{last_name.lower()}@localhost'
    password = bcrypt.generate_password_hash('default')
    name = f'{first_name} {last_name}'
    role = random.choice(roles)
    desc = roles_description[role]
    
    if role == 'SE':
        access_level = 3
    elif role == 'T2':
        access_level = 4
    else:
        access_level = 5
    
    annual_leave = 10

    advocate = {
        'reports_to': reports_to,
        'schedules': schedules,
        'id': id_,
        'staff_id': staff_id,
        'created_at': created_at,
        'updated_at': updated_at,
        'first_name': first_name,
        'last_name': last_name,
        'status': status,
        'email': email,
        'password': password,
        'name': name,
        'role': role,
        'desc': desc,
        'access_level': access_level,
        'annual_leave': annual_leave,
    }
    
    advocates_data.append(advocate)
    print(f'Advocate {i} created successfully.')

print(f'All 50 records created successfully.')

# Convert to DataFrame to visualize
advocates_df = pd.DataFrame(advocates_data)
advocates_df.head()

# Export the generated data to a CSV file
output_file_path = 'C:\\Users\\Owner\\Desktop\\ALX\\staffsync\\generated_advocates_data.csv'
advocates_df.to_csv(output_file_path, index=False)
print('Export to csv successful.')

output_file_path
```
